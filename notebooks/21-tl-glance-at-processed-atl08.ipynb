```python
from pathlib import Path
import pandas as pd
from IPython.display import display

PARQUET_PATH = Path("/scratch/s1803229/ATL08_parquets/S01.parquet")

df = pd.read_parquet(PARQUET_PATH)

print(f"{len(df):,} rows × {len(df.columns)} columns")
if "resource" in df.columns:
    print(f"{df['resource'].nunique():,} overpasses with saved rows")
if "date" in df.columns:
    print(f"Dates: {df['date'].min()} to {df['date'].max()}")

with pd.option_context("display.max_columns", None):
    display(df.head(10))

display(df.dtypes.rename("dtype").to_frame())
```

```python
import pandas as pd
from IPython.display import display

keys = ["site_id", "resource", "date", "lat", "lon"]
derived_columns = []

try:
    for strength in ("strong", "weak"):
        keep = df["beam_str"].eq(strength) & df["outlier"].eq(1)
        for variable in ("Eg", "Ev"):
            column = f"{variable}_{strength}"
            df[column] = df[variable].where(keep)
            derived_columns.append(column)
    del keep

    df_grouped = df.groupby(
        keys, sort=False, observed=True
    ).agg(
        pvpg=("pvpg", "mean"),
        pv=("pv", "max"),
        pg=("pg", "max"),
        Eg_strong=("Eg_strong", "median"),
        Ev_strong=("Ev_strong", "median"),
        Eg_weak=("Eg_weak", "median"),
        Ev_weak=("Ev_weak", "median"),
        data_quantity=("data_quantity", "max"),
        snr=("snr", "mean"),
        layer_flag=("layer_flag", "mean"),
        msw_flag=("msw_flag", "mean"),
        pv_ratio_mean=("pv_ratio_mean", "mean"),
        pv_ratio_max=("pv_ratio_max", "mean"),
        segment_rows=("Eg", "size"),
    ).reset_index()

finally:
    # Remove the temporary radiometry columns from df.
    for column in derived_columns:
        del df[column]

df_grouped["date"] = pd.to_datetime(
    df_grouped["date"], format="%d/%m/%Y", errors="raise"
)

print(f"{len(df_grouped):,} small boxes")
display(df_grouped.head())
```

```python
from pathlib import Path
import pandas as pd

SITE = "S01"
SITES_CSV = Path(
    "../../scratch/climate_site_selection/regional_site_selection/"
    "candidates_d8bbbaa5ea8a7a28/selection_e5c0ae68a48ece02/"
    "n09/selected_sites.csv"
)
UNCERTAINTY_NC = Path(
    "/home/s1803229/Chapter_2/output/classification_uncertainty_EgEv.nc"
)  # Adjust if your saved classifier is elsewhere.

sites = pd.read_csv(SITES_CSV)
site = sites.loc[sites["site_id"].eq(SITE)].iloc[0]
west, south, east, north = map(
    float, (site["west"], site["south"], site["east"], site["north"])
)
```

```python
import numpy as np
import pandas as pd
import xarray as xr

MIN_RATIO = 1.15
MIN_QUANTITY = 13

data = df_grouped.loc[
    df_grouped["Eg_weak"].gt(0)
    & (df_grouped["Eg_strong"] / df_grouped["Eg_weak"]).ge(MIN_RATIO)
    & df_grouped["data_quantity"].ge(MIN_QUANTITY)
    & df_grouped["lon"].between(west, east)
    & df_grouped["lat"].between(south, north)
].dropna(subset=["Eg_strong", "Ev_strong"]).copy()

probability_columns = [
    "prob_no_snow",
    "prob_ground_snow",
    "prob_ground_canopy_snow",
]

with xr.open_dataset(UNCERTAINTY_NC) as model:
    interpolated = model[probability_columns].interp(
        Eg_strong=xr.DataArray(data["Eg_strong"].to_numpy(), dims="observation"),
        Ev_strong=xr.DataArray(data["Ev_strong"].to_numpy(), dims="observation"),
    )
    probabilities = np.column_stack([
        interpolated[name].to_numpy() for name in probability_columns
    ])

valid = np.isfinite(probabilities).all(axis=1)
valid &= probabilities.sum(axis=1) > 0
probabilities[valid] /= probabilities[valid].sum(axis=1, keepdims=True)

data[probability_columns] = probabilities
data["snow_class_id"] = pd.array([pd.NA] * len(data), dtype="Int8")
data["combined_uncertainty"] = np.nan
data.loc[valid, "snow_class_id"] = probabilities[valid].argmax(axis=1)
data.loc[valid, "combined_uncertainty"] = 1 - probabilities[valid].max(axis=1)

print(f"{len(data):,} boxes pass filters; {valid.sum():,} classified")
print(f"{(~valid).sum():,} boxes have unavailable/out-of-range probabilities")
display(data.head())
```

```python
import calendar
import numpy as np
import matplotlib.pyplot as plt
import contextily as cx
from matplotlib.colors import to_rgb
from matplotlib.lines import Line2D
from pyproj import Transformer

YEARS = None                 # None for all, or e.g. [2019, 2020]
POINT_SIZE = 6
UNCERTAINTY_POWER = 1

labels = ["No Snow", "Ground Snow", "Ground + Canopy Snow"]
colors = ["#6b6b6b", "#2878b5", "#e68120"]
palette = np.array([to_rgb(color) for color in colors])

map_data = data.dropna(
    subset=["date", "lat", "lon", "snow_class_id", "combined_uncertainty"]
).copy()
if YEARS is not None:
    map_data = map_data.loc[map_data["date"].dt.year.isin(YEARS)].copy()
if map_data.empty:
    raise ValueError("No classified boxes remain for the selected years.")

transformer = Transformer.from_crs(4326, 3857, always_xy=True)
xmin, ymin = transformer.transform(west, south)
xmax, ymax = transformer.transform(east, north)
map_data["x"], map_data["y"] = transformer.transform(
    map_data["lon"].to_numpy(), map_data["lat"].to_numpy()
)

basemap = cx.providers.OpenStreetMap.Mapnik
map_image, map_extent = cx.bounds2img(
    xmin, ymin, xmax, ymax,
    source=basemap,
    zoom="auto",
    headers={"User-Agent": "ICESat2-Snow-Research/1.0 (University of Edinburgh)"},
    n_connections=1,
    use_cache=True,
)

handles = [
    Line2D([], [], marker="o", linestyle="none", color=color, label=label)
    for label, color in zip(labels, colors)
]
handles.append(
    Line2D([], [], marker="o", linestyle="none",
           markerfacecolor="white", markeredgecolor="grey",
           label="High uncertainty")
)

for year in sorted(map_data["date"].dt.year.unique()):
    year_data = map_data.loc[map_data["date"].dt.year.eq(year)]
    fig, axes = plt.subplots(4, 3, figsize=(15, 18))
    fig.subplots_adjust(bottom=0.07, top=0.94, hspace=0.25, wspace=0.08)

    for month, ax in enumerate(axes.flat, start=1):
        month_data = year_data.loc[
            year_data["date"].dt.month.eq(month)
        ].sort_values("combined_uncertainty")

        ax.imshow(map_image, extent=map_extent, interpolation="bilinear")
        if not month_data.empty:
            base = palette[month_data["snow_class_id"].astype(int).to_numpy()]
            fade = (
                month_data["combined_uncertainty"].clip(0, 2 / 3)
                .to_numpy() / (2 / 3)
            ) ** UNCERTAINTY_POWER
            point_colors = base * (1 - fade[:, None]) + fade[:, None]

            ax.scatter(
                month_data["x"], month_data["y"],
                c=point_colors, s=POINT_SIZE,
                edgecolors="none", rasterized=True,
            )

        # Set limits after imshow/scatter to retain the exact candidate extent.
        ax.set_xlim(xmin, xmax)
        ax.set_ylim(ymin, ymax)
        ax.set_aspect("equal", adjustable="box")
        ax.set_xticks([])
        ax.set_yticks([])
        ax.set_title(
            f"{calendar.month_abbr[month]} {year} | "
            f"{len(month_data):,} boxes"
        )

    fig.suptitle(f"{SITE} — monthly snow classification", fontsize=18)
    fig.legend(handles=handles, loc="lower center", ncol=4, frameon=False)
    fig.text(0.99, 0.005, "© OpenStreetMap contributors",
             ha="right", fontsize=8)
    plt.show()
    plt.close(fig)
```
