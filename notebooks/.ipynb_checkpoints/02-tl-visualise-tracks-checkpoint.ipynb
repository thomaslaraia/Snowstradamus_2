```python
from scripts.track_pairs import *
from scripts.show_tracks import *
from pathlib import Path
from datetime import datetime

def atl08_date_from_path(atl08_path):
    name = Path(atl08_path).name
    datestr = name.split('_')[1][:8]   # YYYYMMDD
    return datetime.strptime(datestr, "%Y%m%d")

dirpaths = ['../data_store/data/sodankyla_full/']
all_coords = [(26.634154,67.361833)]

for dirpath, coords, e in zip(dirpaths, all_coords, elevs):

    all_ATL03, all_ATL08, failed_ATL03 = track_pairs(dirpath, failed=True)
    f = [77]
    date_dt = atl08_date_from_path(all_ATL03[f[0]])
    date_str = date_dt.strftime("%d %B %Y")   # e.g. "04 March 2019"

    big_df = show_tracks(dirpaths[0],[all_ATL03[i] for i in f], [all_ATL08[i] for i in f], coords, c='Eg', CBAR=None,# gtx=[1,3,5],
                         landcover='forest',rebinned=15, sat_flag=1, DW=1, w=4, h=4)

vmin = 0
vmax = 8
plot_static_map_with_box(big_df, coords, c='Eg', cmap='viridis', vmin=vmin, vmax=vmax, w=4, h=4,
                        cam = 'Sodankyla', date_str = date_str
```
