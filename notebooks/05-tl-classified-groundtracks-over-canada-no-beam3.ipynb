```python
df['Eg_strong'] = np.where((df['beam_str'] == 'strong')&(df['outlier'] == 1), df['Eg'], np.nan)
df['Ev_strong'] = np.where((df['beam_str'] == 'strong')&(df['outlier'] == 1), df['Ev'], np.nan)
df['Eg_weak'] = np.where((df['beam_str'] == 'weak')&(df['outlier'] == 1), df['Eg'], np.nan)
df['Ev_weak'] = np.where((df['beam_str'] == 'weak')&(df['outlier'] == 1), df['Ev'], np.nan)
df['forest_type'] = np.where(df['segment_landcover']<119, 'closed', 'open')

df = df[df['beam']!=3]

import warnings

warnings.filterwarnings(
    "ignore",
    message="Mean of empty slice",
    category=RuntimeWarning,
    module="numpy.lib.nanfunctions",
)

df_grouped = df.groupby(['folder','date','lat','lon']).agg({
    'pvpg': 'mean',
    'pv': 'max',
    'pg': 'max',
    'Eg_strong': 'mean',
    'Ev_strong': 'mean',
    'Eg_weak': 'median',
    'Ev_weak': 'median',
    'data_quantity': 'max',
    'snr': 'mean',
    'layer_flag': 'mean',
    'file_index': 'mean',
    'msw_flag': 'mean',
    'pv_ratio_mean': 'mean',
    'pv_ratio_max': 'mean',
    'forest_type': lambda x: pd.Series.mode(x).iloc[0],
}).reset_index()

df_grouped = df_grouped.dropna(subset=['Eg_strong', 'Ev_strong'])

data = df_grouped[(df_grouped['Eg_strong']/df_grouped['Eg_weak'] >= 1.15)
    &(df_grouped['data_quantity'] >= 26)]
```

```python
import numpy as np
import pandas as pd
import xarray as xr


UNCERTAINTY_NC = "../output/classification_uncertainty_EgEv_no_beam3.nc"

probability_variables = [
    "prob_no_snow",
    "prob_ground_snow",
    "prob_ground_canopy_snow",
]

class_names = {
    0: "No Snow",
    1: "Ground Snow",
    2: "Ground + Canopy Snow",
}

eg = xr.DataArray(
    data["Eg_strong"].to_numpy(),
    dims="observation",
)
ev = xr.DataArray(
    data["Ev_strong"].to_numpy(),
    dims="observation",
)

with xr.open_dataset(UNCERTAINTY_NC) as uncertainty_ds:
    interpolated = uncertainty_ds[probability_variables].interp(
        Eg_strong=eg,
        Ev_strong=ev,
    )

    probabilities = np.column_stack([
        interpolated[name].to_numpy()
        for name in probability_variables
    ])

# Values outside the NetCDF's Eg/Ev range remain NaN rather than being clipped.
valid = np.isfinite(probabilities).all(axis=1)
probabilities[valid] /= probabilities[valid].sum(axis=1, keepdims=True)

data = data.copy()
data[probability_variables] = probabilities

data["snow_class_id"] = pd.Series(pd.NA, index=data.index, dtype="Int8")
data["snow_class"] = pd.NA
data["predicted_class_probability"] = np.nan
data["combined_uncertainty"] = np.nan

predicted_class = probabilities[valid].argmax(axis=1)
predicted_probability = probabilities[valid].max(axis=1)

data.loc[valid, "snow_class_id"] = predicted_class
data.loc[valid, "snow_class"] = [class_names[k] for k in predicted_class]
data.loc[valid, "predicted_class_probability"] = predicted_probability
data.loc[valid, "combined_uncertainty"] = 1 - predicted_probability

print(f"Classified rows: {valid.sum():,}/{len(data):,}")
print(f"Rows outside the NetCDF grid: {(~valid).sum():,}")

display(
    data[[
        "Eg_strong",
        "Ev_strong",
        "snow_class",
        "predicted_class_probability",
        "combined_uncertainty",
    ]].head()
)
```

```python
import calendar
import contextily as cx
import matplotlib.pyplot as plt

from matplotlib.colors import to_rgb
from matplotlib.lines import Line2D
from pyproj import Transformer


CLASS_COLORS = {
    "No Snow": "#6b6b6b",
    "Ground Snow": "#2878b5",
    "Ground + Canopy Snow": "#e68120",
}

BASEMAP = cx.providers.OpenStreetMap.Mapnik
TILE_HEADERS = {
    "User-Agent": "ICESat2-Snow-Research/1.0 (University of Edinburgh)",
}
HIGH_UNCERTAINTY_COLOR = "#ffffff" #"#e600ff"
UNCERTAINTY_POWER = 1

# Change these if you only want one folder or selected years.
FOLDER = None
YEARS = None
POINT_SIZE = 10

map_data = data.dropna(
    subset=["date", "lat", "lon", "snow_class", "combined_uncertainty"]
).copy()

map_data["date"] = pd.to_datetime(
    map_data["date"],
    dayfirst=True,
    errors="coerce",
)
map_data = map_data.dropna(subset=["date"])

if FOLDER is not None:
    map_data = map_data[map_data["folder"] == FOLDER]

if YEARS is not None:
    map_data = map_data[
        map_data["date"].dt.year.isin(np.atleast_1d(YEARS))
    ]

if map_data.empty:
    raise ValueError("No observations remain after filtering.")

transformer = Transformer.from_crs(
    "EPSG:4326",
    "EPSG:3857",
    always_xy=True,
)
map_data["x"], map_data["y"] = transformer.transform(
    map_data["lon"].to_numpy(),
    map_data["lat"].to_numpy(),
)

map_data["year"] = map_data["date"].dt.year
map_data["month"] = map_data["date"].dt.month

legend_handles = [
    Line2D(
        [0], [0],
        marker="o",
        linestyle="none",
        markerfacecolor=color,
        markeredgecolor="none",
        markersize=7,
        label=label,
    )
    for label, color in CLASS_COLORS.items()
]
legend_handles += [
    Line2D(
        [0], [0],
        marker="o",
        linestyle="none",
        markerfacecolor=HIGH_UNCERTAINTY_COLOR,
        markeredgecolor="#666666",
        markersize=7,
        label="High uncertainty",
    ),
]

for year, year_data in map_data.groupby("year", sort=True):
    xmin, xmax = year_data["x"].min(), year_data["x"].max()
    ymin, ymax = year_data["y"].min(), year_data["y"].max()
    xpad = max((xmax - xmin) * 0.06, 10_000)
    ypad = max((ymax - ymin) * 0.06, 10_000)

    map_image, map_extent = cx.bounds2img(
        xmin - xpad,
        ymin - ypad,
        xmax + xpad,
        ymax + ypad,
        source=BASEMAP,
        zoom="auto",
        headers=TILE_HEADERS,
        n_connections=1,
        use_cache=True,
    )

    fig, axes = plt.subplots(
        4,
        3,
        figsize=(15, 17),
        constrained_layout=True,
    )

    for month, ax in enumerate(axes.ravel(), start=1):
        month_data = year_data[year_data["month"] == month]

        if month_data.empty:
            ax.set_title(f"{calendar.month_abbr[month]} {year}\nNo data")
            ax.axis("off")
            continue

        ax.set_xlim(xmin - xpad, xmax + xpad)
        ax.set_ylim(ymin - ypad, ymax + ypad)

        ax.imshow(
            map_image,
            extent=map_extent,
            interpolation="bilinear",
            zorder=0,
        )
        cx.add_attribution(ax, BASEMAP.attribution, font_size=5)

        # Plot high-uncertainty points last so they remain visible.
        month_data = month_data.sort_values(
            "combined_uncertainty",
            ascending=True,
        )

        base_colors = np.array([
            to_rgb(CLASS_COLORS[snow_class])
            for snow_class in month_data["snow_class"]
        ])
        uncertainty_color = np.array(to_rgb(HIGH_UNCERTAINTY_COLOR))

        uncertainty_strength = (
            month_data["combined_uncertainty"].clip(0, 2 / 3) / (2 / 3)
        ).to_numpy() ** UNCERTAINTY_POWER

        point_colors = (
            (1 - uncertainty_strength[:, None]) * base_colors
            + uncertainty_strength[:, None] * uncertainty_color
        )

        ax.scatter(
            month_data["x"],
            month_data["y"],
            s=POINT_SIZE,
            c=point_colors,
            edgecolors="none",
            rasterized=True,
            zorder=3,
        )

        ax.set_title(
            f"{calendar.month_abbr[month]} {year}\n"
            f"n = {len(month_data):,}"
        )
        ax.set_xticks([])
        ax.set_yticks([])

    title = f"ICESat-2 snow classification: {year}"
    if FOLDER is not None:
        title += f" — {FOLDER}"

    fig.suptitle(title, fontsize=17)
    fig.legend(
        handles=legend_handles,
        loc="lower center",
        ncol=4,
        frameon=False,
        bbox_to_anchor=(0.5, -0.015),
    )

    plt.show()
```
