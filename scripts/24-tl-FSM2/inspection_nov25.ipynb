```python
import numpy as np
import pandas as pd
from IPython.display import display

SELECT_BY = "sf"       # "tp": wettest period; "sf": snowiest period.
TOL = 1e-7            # Metres water equivalent; matches the runner.

time = "valid_time" if "valid_time" in ds.dims else "time"

# Choose a valid forest cell with substantial precipitation.
eligible = (
    np.isfinite(ds.VAI) & np.isfinite(ds.vegh)
    & (ds.VAI > 0) & (ds.vegh > 2)
)
score = (
    ds[SELECT_BY].max(time).where(eligible)
    .transpose("latitude", "longitude").compute()
)
if not np.isfinite(score.values).any() or float(score.max()) <= 0:
    raise ValueError("No precipitation found at eligible cells; try another month.")

iy, ix = np.unravel_index(np.nanargmax(score.values), score.shape)

point = ds[["tp", "sf"]].isel(latitude=iy, longitude=ix).compute()
stored = point.to_dataframe()[["tp", "sf"]].sort_index()

if stored.index.has_duplicates:
    raise ValueError("Duplicate timestamps.")

# Assign 01:00 ... next-day 00:00 to the same forecast cycle.
cycles = []
for day, group in stored.groupby(
    (stored.index - pd.Timedelta(hours=1)).normalize()
):
    expected = pd.date_range(day + pd.Timedelta(hours=1), periods=24, freq="h")
    if group.index.equals(expected) and np.isfinite(group.to_numpy()).all():
        cycles.append((day, group))

if not cycles:
    raise ValueError("No complete, finite 24-hour forecast cycle found.")

day, raw = max(cycles, key=lambda item: item[1][SELECT_BY].max())

# Test the cumulative interpretation: 01 UTC is the first hourly amount.
increments = raw.diff()
increments.iloc[0] = raw.iloc[0]

table = pd.DataFrame({
    "Stored tp (mm)": raw.tp * 1000,
    "Hourly tp (mm)": increments.tp * 1000,
    "Stored sf (mm SWE)": raw.sf * 1000,
    "Hourly sf (mm SWE)": increments.sf * 1000,
})
print(f"Cell: {float(point.latitude):.3f}, {float(point.longitude):.3f}")
print(f"Forecast cycle: {day:%Y-%m-%d} 01 UTC → next day 00 UTC")
display(table.round(6))

summary = pd.DataFrame({
    "Sum of hourly increments (mm)": increments.sum() * 1000,
    "Stored ending 00 UTC total (mm)": raw.iloc[-1] * 1000,
    "Difference (mm)": (increments.sum() - raw.iloc[-1]) * 1000,
    "Minimum hourly increment (mm)": increments.min() * 1000,
})
display(summary)

negative = increments < -TOL
rain_negative = (increments.tp - increments.sf) < -TOL

if negative.to_numpy().any():
    print("STOP: significant negative increments under the cumulative interpretation:")
    display((increments * 1000).where(negative).dropna(how="all"))
    print("Investigate the accumulation convention; these values have NOT been clipped.")
elif rain_negative.any():
    print("STOP: hourly snowfall exceeds total precipitation beyond tolerance.")
else:
    print("PASS: no significant negative increments; snowfall does not exceed total precipitation.")
    if (increments.to_numpy() < 0).any():
        print("Tiny negative increments exist within the runner's rounding tolerance.")
```

```python
from pathlib import Path
import pandas as pd
import xarray as xr

root = Path("/scratch/s1803229/fsm2/S01")
parts = []

for path in sorted(root.glob("????_??/driving_data.nc")):
    with xr.open_dataset(path) as ds:
        parts.append(ds["ssrd"].isel(latitude=0, longitude=0).to_series())

ssrd = pd.concat(parts).astype("float64")
if not ssrd.index.to_series().diff().iloc[1:].eq(pd.Timedelta(hours=1)).all():
    raise ValueError("Time axis has a gap, duplicate, or out-of-order hour.")

# Match the runner: 01 UTC starts a new accumulation cycle.
increment = ssrd.diff()
reset = ssrd.index.hour == 1
increment.loc[reset] = ssrd.loc[reset]

bad = increment[increment < -1.0]
print(f"Hourly increments below −1 J/m²: {len(bad):,}")
if not bad.empty:
    t = bad.idxmin()
    print(f"Largest decrease: {bad.loc[t]:.4f} J/m² at {t} UTC")
    print("Crosses monthly file boundary:", t.day == 1 and t.hour == 0)
    window = pd.concat(
        {"stored_J_m2": ssrd, "increment_J_m2": increment}, axis=1
    )
    print(window.loc[t - pd.Timedelta(hours=3):t + pd.Timedelta(hours=3)])
```

```python
from pathlib import Path

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import xarray as xr
from IPython.display import HTML, display
from matplotlib.animation import FuncAnimation

experiment = Path("/scratch/s1803229/fsm2_ensembles/S01/nov25")
region = "S01"
members = {
    "CLM5": "clm5_cap1_loss1",
    "Noah-MP": "noahmp_cap1_loss1",
    "HP98 spruce": "hp98_spruce_cap1_loss1",
    "HP98 pine": "hp98_pine_cap1_loss1",
}

if not experiment.is_dir():
    raise FileNotFoundError(experiment)

regional = {}
cell_files = {}
for label, member in members.items():
    folder = experiment / member
    regional[label] = folder / f"{region}.nc"
    cell_files[label] = {
        p.name: p for p in (folder / region).glob("cell_y*_x*.nc")
        if not p.name.endswith(".partial.nc")
    }

common_cells = sorted(set.intersection(*(set(files) for files in cell_files.values())))

display(pd.DataFrame({
    "Completed cell files": {label: len(files) for label, files in cell_files.items()},
    "Regional file ready": {label: path.is_file() for label, path in regional.items()},
}))
print(f"Cells available in all four runs: {len(common_cells)}")
```

```python
if not common_cells:
    raise ValueError("No cell has finished in all four runs yet. Rerun cell 11 later.")

cell = common_cells[0]
winter = slice("2020-10-01", "2021-05-31")
variables = ["canopy_snow_mass", "ground_snow_mass", "snow_depth"]

fig, axes = plt.subplots(3, 1, figsize=(12, 8), sharex=True, constrained_layout=True)

for label, files in cell_files.items():
    with xr.open_dataset(files[cell]) as ds:
        snow = ds[variables].sel(time=winter).load()
        lat, lon = float(ds.latitude), float(ds.longitude)

    ground_cover = 100 * (snow.snow_depth / 0.10).clip(0, 1)
    for ax, field in zip(axes, [snow.canopy_snow_mass, snow.ground_snow_mass, ground_cover]):
        ax.plot(snow.time, field, label=label, linewidth=1.2, alpha=0.85)

for ax, ylabel in zip(axes, ["Canopy snow (kg m⁻²)", "Ground snow (kg m⁻²)", "Ground cover (%)"]):
    ax.set_ylabel(ylabel)

axes[0].legend(ncol=4)
axes[2].set_ylim(-2, 102)
fig.suptitle(f"{region} · {cell} · {lat:.2f}°, {lon:.2f}°")
plt.show()
```

```python
when = "2021-02-15T12:00"
variable = "canopy_snow_mass"
# Alternatives: "ground_snow_mass", "snow_depth", "ground_snow_cover_fraction"
cmap = "Blues"

missing = [label for label, path in regional.items() if not path.is_file()]
if missing:
    raise FileNotFoundError(f"Waiting for assembled regional files: {', '.join(missing)}")

maps = {}
for label, path in regional.items():
    with xr.open_dataset(path) as ds:
        maps[label] = ds[variable].sel(time=when).load()

if variable == "ground_snow_cover_fraction":
    maps = {label: field * 100 for label, field in maps.items()}
    vmax, units = 100, "%"
else:
    vmax = max(float(field.max()) for field in maps.values()) or 1
    units = next(iter(maps.values())).attrs.get("units", "")

fig, axes = plt.subplots(2, 2, figsize=(10, 8), sharex=True, sharey=True,
                         constrained_layout=True)

for ax, (label, field) in zip(axes.flat, maps.items()):
    mesh = field.plot.pcolormesh(
        ax=ax, x="longitude", y="latitude", cmap=cmap,
        vmin=0, vmax=vmax, add_colorbar=False,
    )
    ax.set(title=label, facecolor="lightgrey")

fig.colorbar(mesh, ax=axes.ravel().tolist(), label=units, shrink=0.8)
fig.suptitle(f"{region} · {variable.replace('_', ' ')} · {when.replace('T', ' ')} UTC")
plt.show()
```

```python
model = "HP98 spruce"  # "CLM5", "Noah-MP", "HP98 spruce", or "HP98 pine"
start, end = "2020-10-01", "2021-05-31"
frequency = "7D"
fps = 4

path = regional[model]
if not path.is_file():
    raise FileNotFoundError(f"Regional file for {model} is not ready: {path}")

with xr.open_dataset(path) as ds:
    dates = pd.date_range(start, end, freq=frequency)
    frames = ds[["canopy_snow_mass", "ground_snow_cover_fraction"]].sel(time=dates).load()

mass = frames.canopy_snow_mass
layers = [
    ((mass > 0).astype(float).where(mass.notnull()),
     "Canopy snow present", "Oranges", 1),
    (100 * frames.ground_snow_cover_fraction,
     "Ground snow cover (%)", "Blues", 100),
]

fig, axes = plt.subplots(1, 2, figsize=(10, 4.5), constrained_layout=True)
meshes = []
for ax, (field, title, cmap, vmax) in zip(axes, layers):
    mesh = field.isel(time=0).plot.pcolormesh(
        ax=ax, x="longitude", y="latitude", cmap=cmap,
        vmin=0, vmax=vmax, add_colorbar=False,
    )
    ax.set(title=title, facecolor="lightgrey")
    bar = fig.colorbar(mesh, ax=ax)
    if vmax == 1:
        bar.set_ticks([0, 1], labels=["Absent", "Present"])
    meshes.append(mesh)

heading = fig.suptitle("")

def update(i):
    for mesh, (field, *_) in zip(meshes, layers):
        mesh.set_array(field.isel(time=i).values.ravel())
    heading.set_text(f"{region} · {model} · {dates[i]:%Y-%m-%d %H:%M} UTC")

animation = FuncAnimation(fig, update, frames=len(dates), interval=1000 / fps)
plt.close(fig)
display(HTML(animation.to_jshtml()))
```
