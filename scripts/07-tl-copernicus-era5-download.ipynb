```python
from pathlib import Path
import zipfile
import calendar
import cdsapi
import xarray as xr
import shutil

year = "2025"
month = "03"

out_dir = Path(r"C:/Users/s1803229/Documents/Snowstradamus/scratch/ERA5")
extract_dir = out_dir / f"{year}_{month}"
extract_dir.mkdir(parents=True, exist_ok=True)

dataset = "reanalysis-era5-land"
client = cdsapi.Client()

days = [f"{d:02d}" for d in range(1, calendar.monthrange(int(year), int(month))[1] + 1)]
times = [f"{h:02d}:00" for h in range(24)]

base_request = {
    "year": year,
    "month": month,
    "day": days,
    "time": times,
    "data_format": "netcdf",
    "download_format": "zip",
    "area": [53, -95, 48, -87],
}

variable_groups = {
    "temp": [
        "2m_dewpoint_temperature",
        "2m_temperature",
    ],
    "met": [
        "snowfall",
        "surface_solar_radiation_downwards",
        "surface_thermal_radiation_downwards",
        "10m_u_component_of_wind",
        "10m_v_component_of_wind",
        "surface_pressure",
        "total_precipitation",
    ],
}

part_paths = []

for name, variables in variable_groups.items():
    request = base_request.copy()
    request["variable"] = variables

    zip_path = extract_dir / f"{year}_{month}_{name}.zip"
    client.retrieve(dataset, request, str(zip_path))

    part_dir = extract_dir / name
    part_dir.mkdir(exist_ok=True)

    with zipfile.ZipFile(zip_path, "r") as z:
        z.extractall(part_dir)

    nc_files = list(part_dir.rglob("*.nc"))
    if len(nc_files) != 1:
        raise ValueError(f"{name}: found {len(nc_files)} .nc files")

    part_nc = extract_dir / f"{year}_{month}_{name}.nc"
    nc_files[0].replace(part_nc)
    part_paths.append(part_nc)

    zip_path.unlink()

datasets = []

for p in part_paths:
    with xr.open_dataset(p) as d:
        datasets.append(d.load())   # load into memory so files can be deleted

ds = xr.merge(datasets, compat="override")

final_path = extract_dir / f"{year}_{month}.nc"

if final_path.exists():
    final_path.unlink()

ds.to_netcdf(final_path)
ds.close()

# Delete temporary part .nc files
for p in part_paths:
    p.unlink(missing_ok=True)

# Delete temporary extracted folders
for name in variable_groups:
    shutil.rmtree(extract_dir / name, ignore_errors=True)
```
